```sql
CREATE CATALOG IF NOT EXISTS wanderbricks_analytics;

CREATE SCHEMA IF NOT EXISTS wanderbricks_analytics.analytics;

CREATE OR REPLACE VIEW wanderbricks_analytics.analytics.property_marketing AS
SELECT 
    v.view_id,
    v.timestamp AS view_timestamp,
    v.device_type,
    p.property_id,
    v.referrer,
    v.user_id,
    p.host_id,
    p.destination_id,
    p.base_price,
    p.property_type 
FROM samples.wanderbricks.page_views v
JOIN samples.wanderbricks.properties p
ON v.property_id = p.property_id;
```
