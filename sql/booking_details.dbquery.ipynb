```sql
CREATE CATALOG IF NOT EXISTS wanderbricks_analytics;

CREATE SCHEMA IF NOT EXISTS wanderbricks_analytics.analytics;

CREATE OR REPLACE VIEW wanderbricks_analytics.analytics.booking_details AS
SELECT
    b.booking_id,
    b.created_at AS booking_created_at,
    b.updated_at AS booking_updated_at,
    b.check_in,
    b.check_out,
    DATEDIFF(DAY, b.check_in, b.check_out) AS stay_length,
    b.guests_count,
    b.total_amount,
    b.status,
    u.user_id,
    u.country AS user_country,
    u.user_type,
    p.property_id,
    p.property_type,
    p.base_price,
    p.max_guests,
    p.bedrooms,
    p.bathrooms,
    p.destination_id,
    d.destination AS destination_name,
    d.country AS destination_country,
    h.host_id,
    h.rating AS host_rating,
    h.country AS host_country,
    h.is_active,
    h.is_verified
FROM samples.wanderbricks.bookings b
JOIN samples.wanderbricks.properties p
    ON b.property_id = p.property_id
JOIN samples.wanderbricks.users u
    ON b.user_id = u.user_id
LEFT JOIN samples.wanderbricks.destinations d
    ON p.destination_id = d.destination_id
LEFT JOIN samples.wanderbricks.hosts h
    ON p.host_id = h.host_id;
```
