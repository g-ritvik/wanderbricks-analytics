# Database Schema

```text
users
   │
   │ user_id
   ▼
bookings ───────────────► properties
   │                        │
   │ property_id            │ host_id
   ▼                        ▼
booking_updates          hosts

properties
      │
      ├────────► property_amenities ─────► amenities
      │
      ├────────► property_images
      │
      └────────► page_views
```

## Relationships

- **users**
  - `user_id` → **bookings**

- **bookings**
  - `property_id` → **properties**
  - One-to-many → **booking_updates**

- **properties**
  - `host_id` → **hosts**
  - One-to-many → **property_amenities**
  - One-to-many → **property_images**
  - One-to-many → **page_views**

- **property_amenities**
  - References **properties**
  - References **amenities**
